```python
import math
import numpy as np
import pandas as pd
```

## Método de la bisección

```python
def Biseccion(f,a,b,tol,maxiter):
    iters = 0
    error = tol+1

    while error>tol and iters<=maxiter:
        x = (a+b)/2
        error = (b-a)/2
        if f(a)*f(x)>0:
            a = x
        else:
            b = x
        iters = iters+1

    return [x,iters,error]
```

## Función crea una tabla con los resultados de las iteraciones

```python
def Bisecciontabla(f,a,b,tol,maxiter):
    iters = 0
    error = tol+1
    lista_a = list(); lista_b = list()
    lista_a.append(a); lista_b.append(b);
    lista_x = list(); lista_e = list()

    while error>tol and iters<=maxiter:
        x = (a+b)/2
        error = (b-a)/2
        lista_e.append(error); lista_x.append(x)
        if f(a)*f(x)>0:
            a = x
        else:
            b = x
        iters = iters+1
        lista_a.append(a); lista_b.append(b);
    lista_a.pop(); lista_b.pop()
    tabla = pd.DataFrame({'a': lista_a, 'b': lista_b, 'x': lista_x, 'Errores': lista_e})
    pd.set_option('display.precision', 10)
    return tabla
```

### Ejemplo 

Aplicar el método de la bisección para la ecuación $\sqrt{x}-\cos (x)=0$ en el intervalo $[0,1]$ con una tolerancia de $10^{-3}$

```python
#Definimos f

f = lambda x: math.sqrt(x)-math.cos(x)
a = 0
b = 1
tol = 1e-3
maxiter = 30

##Llamamos a la funcion
x,iters,error = Biseccion(f,a,b,tol,maxiter)
```

```python
print("La solucion obtenida fue x=%s" %x)
print("La cantidad de iteraciones realizadas fueron %s" %iters)
print("El error que se obtuvo fue de %s" %error)
```

### Mostrar la tabla con todas las iteraciones

```python
Bisecciontabla(f,a,b,tol,maxiter)
```

## Ejemplo

Encontrar la solucion de la ecuacion $e^{-x}(3.2 \sin(x)-0.5 \cos(x))=0$ 
en el intervalo $[3,4]$ con una tolerancia de $10^{-4}$

```python
## Introducir f

f = lambda x: math.exp(-x)*(3.2*math.sin(x)-0.5*math.cos(x))
a = 3
b = 4
tol = 1e-4
maxiter = 30
```

```python
x,iters,error = Biseccion(f,a,b,tol,maxiter)
```

```python
print("La solucion obtenida fue x=%s" %x)
print("La cantidad de iteraciones realizadas fueron %s" %iters)
print("El error que se obtuvo fue de %s" %error)
```

```python
Bisecciontabla(f,a,b,tol,maxiter)
```
