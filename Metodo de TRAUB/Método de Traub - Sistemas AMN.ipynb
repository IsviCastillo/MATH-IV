# Método de Traub para sistemas de ecuaciones no lineales

```python
import math
import numpy as np
import pandas as pd
```

## Función que implementa el método de Traub

```python
def TraubSis(F,JF,x0,tol,maxiter):
    iters = 0 #Contador de iteraciones a cero
    error = tol+1
    while error>tol and iters<=maxiter:
        z = np.linalg.solve(JF(x0),F(x0)) #Se resuelve el sistema lineal
        y = x0-z
        w = np.linalg.solve(JF(x0),F(y)) #Se resuelve el sistema lineal
        x = y-w
        error = np.linalg.norm(x-x0) #Se encuentra el error con la norma
        x0 = x
        iters = iters+1
    return [x,iters,error]
```

## Función que muestra la tabla del método de Traub

```python
def TraubSisTabla(F,JF,x0,tol,maxiter):
    iters = 0 #Contador de iteraciones a cero
    error = tol+1
    lista_error = list()
    lista_x = list()
    while error>tol and iters<=maxiter:
        z = np.linalg.solve(JF(x0),F(x0)) #Se resuelve el sistema lineal
        y = x0-z
        w = np.linalg.solve(JF(x0),F(y)) #Se resuelve el sistema lineal
        x = y-w
        lista_x.append(x)
        error = np.linalg.norm(x-x0) #Se encuentra el error con la norma
        lista_error.append(error)
        x0 = x
        iters = iters+1
    Tabla = pd.DataFrame()
    Tabla["x"] = [w[0] for w in lista_x]
    Tabla["y"] = [w[1] for w in lista_x]
    Tabla["Errores"] = lista_error
    pd.set_option('display.precision', 16)
    return Tabla
```

### Ejemplo 1

Encontrar la solución al sistemas de ecuaciones
\begin{align*}
x^2-2x-y+1/2&=0 \\
x^2+4y^2-4&=0
\end{align*}

Con el método de Traub usando las estimaciones iniciales $x_0=1$ e $y_0=1$ y una tolerancia de $10^{-10}$

```python
def F(w):
    x,y = w
    S = np.zeros(2)
    S[0] = x**2-2*x-y+1/2
    S[1] = x**2+4*y**2-4
    return S

#Matriz Jacobiana
def JF(w):
    x,y = w
    M = np.zeros([2,2])
    M[0,0] = 2*x-2; M[0,1] = -1
    M[1,0] = 2*x; M[1,1] = 8*y
    return M

x0 = np.array([1,1])
tol = 1e-10
maxiter = 30

x,iters,error = TraubSis(F,JF,x0,tol,maxiter)
```

```python
print("La solución al sistemas de ecuaciones es %s" %x)
print("La cantidad de iteraciones realizadas fueron %s"%iters)
print("El error que se obtuvo fue de %s" %error)
```

```python
TraubSisTabla(F,JF,x0,tol,maxiter)
```

## Ejemplo 2

Resolver el siguiente sistema no lineal
\begin{align*}
4x^2-20x+\frac{1}{4}y^2+8 &=0 \\
\dfrac{1}{2}xy^2+2x-5y+8&=0
\end{align*}
Con una tolerancia de $10^{-6}$ y una estimación inicial de $\mathbf{x}^{(0)}=(0,0)$

```python
def F(w):
    x,y = w
    S = np.zeros(2)
    S[0] = 4*x**2-20*x+1/4*y**2+8
    S[1] = 1/2*x*y**2+2*x-5*y+8
    return S

#Matriz Jacobiana
def JF(w):
    x,y = w
    M = np.zeros([2,2])
    M[0,0] = 8*x-20; M[0,1] = 1/2*y
    M[1,0] = 1/2*y**2+2; M[1,1] = x*y-5
    return M

x0 = np.array([0,0])
tol = 1e-10
maxiter = 30

x,iters,error = TraubSis(F,JF,x0,tol,maxiter)
print("La solución al sistema de ecuaciones es %s" %x)
print("La cantidad de iteraciones realizadas fueron %s"%iters)
print("El error que se obtuvo fue de %s" %error)
```

```python
TraubSisTabla(F,JF,x0,tol,maxiter)
```
